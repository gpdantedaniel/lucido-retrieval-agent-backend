```python
DESCRIPTION_TEMPLATE="""title: {title}
scientific name: {scientific_name}
located: {coordinates}
information: 
{information}
"""

def describe_artifact(row):
    """
    Generates a description of a row
    by formatting its values of interest
    """

    title = row['Title EN']
    scientific_name = row['Scientific Name EN']
    coordinates = row['Museum \nCollection number']

    # There has to be a better way of doing this
    questions = row[['Question 1 EN', 'Question 2 EN', 'Question 3 EN', 'Question 4 EN', 'Question 5 EN', 'Question 6 EN']].dropna().to_list()
    facts = row[['Fact 1 EN', 'Fact 2 EN', 'Fact 3 EN','Fact 4 EN', 'Fact 5 EN', 'Fact 6 EN']].dropna().to_list()

    information = '\n'.join(f'{question} {fact}' for question, fact in zip(questions, facts))

    description = DESCRIPTION_TEMPLATE.format(
        title=title,
        scientific_name=scientific_name,
        coordinates=coordinates,
        information=information
    )

    return title, description
```

```python
title, description = describe_artifact(df.iloc[10])
print(description)
```

```python
formatted = pd.DataFrame(df.dropna(subset=['Title EN']).apply(describe_artifact, axis=1, result_type='expand'))
formatted = formatted.rename({0: 'title', 1: 'content'}, axis=1)

formatted
```

```python
formatted.to_csv('../../data/processed/vrmuseum.csv')
```
