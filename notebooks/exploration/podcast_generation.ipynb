```python
from kokoro import KPipeline, KModel
from IPython.display import display, Audio
import soundfile as sf
import numpy as np
import torch
from io import BytesIO

pipeline = KPipeline(lang_code='a', repo_id='hexgrad/Kokoro-82M')

text = """Welcome to today's podcast: Chemical adaptations in ants.——

Ants rely heavily on chemical communication and adaptations for survival and social coordination. 
They use pheromones—volatile chemical signals—to mark trails, warn of threats, and organize colony activities, functioning as a chemical language.
Their exoskeletons, made primarily of chitin and reinforced with minerals and proteins, provide protection and flexibility, enabling ants 
to perform tasks like foraging and nest building. Additionally, ants produce venom for defense and prey immobilization, and some species 
have metapleural glands that secrete antimicrobial substances, protecting the colony from pathogens. These chemical strategies, 
along with their ability to modify soil and form mutualistic relationships (e.g., with fungi in leaf-cutting ants), 
highlight ants' significant ecological impact and their role in maintaining ecosystem health. Their chemical adaptations 
demonstrate both evolutionary innovation and environmental influence.
"""

generator = pipeline(text, voice='af_heart', split_pattern=None)
audio = np.concatenate([segment[-1] for segment in generator])

buffer = BytesIO()

sf.write(buffer, audio, 24_000, format='WAV', subtype='PCM_24')
```
